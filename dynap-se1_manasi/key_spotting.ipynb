```python
model.update_parameter_group(param_group, 0, 0)

# Set up a population
N_POP = 4 # 4 bands
NE_PER_POP = 20 # neurons per band
# NI_POP = 4 # There should always be same number of e and i populations
NI_PER_POP = 20
CHIP = 0
#NI = 40

# experiment duration
stim_time = 3
post_stim_time = 3
total_time = stim_time+post_stim_time

# probabilities
p_INP_E = 0.8
p_E_E   = 0.05
p_I_I   = 0.1
p_E_I   = 0.5
p_I_E   = 0.6

# Create a NetworkGenerator object to build the network configuration:
net_gen = NetworkGenerator()
net_gen.clear_network() # Clear any existing network structure

# spike generator in chip=0, core=1, neuron=1
spikegen_ids = [(0,i,NE_PER_POP+NI_PER_POP) for i in range(N_POP)]
# 1 spikegen pro population at position after all excit and inhib neurons
# for core in range(N_POP):
#     spikegen_ids.append((chip, core, neuron) for neuron in range(NE_PER_POP))
    
spikegens = []
for spikegen_id in spikegen_ids:
    spikegens.append(Neuron(spikegen_id[0],spikegen_id[1],spikegen_id[2],True))
    

# excitatory population
pop_E = []

for core_id in range(N_POP):    
    # core_id=N_POP, ni=N_PER_POP+1; e.g. [(0,0,0), (0,0,1), ..., (0,3,18), (0,3,19)]
    pop_ids = [(0, core_id, ni) for ni in range(NE_PER_POP)]
    pop = [Neuron(nid[0], nid[1], nid[2]) for nid in pop_ids]
    # [ [ (0,0,0),(),...,(0,0,19) ] , [ (0,1,0),...,(0,1,19) ] , [ (0,2,0),...,(0,2,19) ] , [ (0,3,0),...,(0,3,19) ] ]
    pop_E.append(pop)

# inhibitory population
pop_I = []

for core_id in range(N_POP):
    # e.g. [(0,0,20), (0,0,21), ..., (0,3,38), (0,3,39)]
    pop_ids = [(0,core_id, ni) for ni in range(NE_PER_POP, NE_PER_POP+NI_PER_POP)]
    pop = [Neuron(nid[0], nid[1], nid[2]) for nid in pop_ids]
    pop_I.append(pop)


# ==================================================================================
# Connections for Winner-Take-All

# connect SPIKEGEN to POP_E
for index in range(len(pop_E)):
    for i in range(int(NE_PER_POP*p_INP_E)+1):
        # Here, we connect from spikegens[0] to neurons_E[i]
        # using the AMPA synapse type
        net_gen.add_connection(spikegens[index], pop_E[index][i], dyn1.Dynapse1SynType.AMPA)

# connect POPULATION E to POPULATION I
for idx_e, pop_e in enumerate(pop_E):
    for idx_i, pop_i in enumerate(pop_I):
        if idx_e == idx_i:
            for i in range(NE_PER_POP):
                for j in range(NI_PER_POP):
                    if(p_E_I > np.random.uniform()):
                        net_gen.add_connection(pop_e[i], pop_i[j], dyn1.Dynapse1SynType.AMPA)

# connect POPULATION I to POPULATION E
for idx_i, pop_i in enumerate(pop_I):
    for idx_e, pop_e in enumerate(pop_E):
        if idx_i!=idx_e:
            for i in range(NI_PER_POP):
                for j in range(NE_PER_POP):
                    if(p_I_E > np.random.uniform()):
                        # how to make this negative???
                        net_gen.add_connection(pop_i[i], pop_e[j], dyn1.Dynapse1SynType.GABA_B)

# ===================================================================================
```
